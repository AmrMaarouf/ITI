```python
import numpy as np
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
```

```python
print("Matrix A:", "\n", A, "\n")
print("Matrix B:", "\n", B)
```

```python
print("ِAddition (A + B):")
print(A+B)
```

```python
print("ِDot Product (A @ B):")
print(np.dot(A,B))

#print(A @ B)
```

```python
print("Transpose of A:")
print(np.transpose(A))
```

```python
A_mat = np.array([[2, 3],
                  [3, 1]])
B_mat = np.array([8, 5])

solution = np.linalg.solve(A_mat, B_mat)

print(f"The solution is: x = {solution[0]:.2f}, y = {solution[1]:.2f}")
```
