# Problem 1

Use numpy to simulate the vector outcomes for repeating the following random experiments $1000$ times and use the relative frequency to compute the probability of the corresponding events:
    
1. **Experiment:** Flipping a fair coin, **Event:** A head occurs.
     
Observe how changing the number of repetition affects the computed probabilities.

**Hints:**
- Use [numpy.random.randint](https://numpy.org/doc/stable/reference/random/generated/numpy.random.randint.html) or [numpy.random.choice](https://numpy.org/doc/stable/reference/random/generated/numpy.random.choice.html) to simulate the experiment.
- Other useful functions [numpy.arange](https://numpy.org/doc/stable/reference/generated/numpy.arange.html), [numpy.cumsum](https://numpy.org/doc/stable/reference/generated/numpy.cumsum.html), [numpy.sum](https://numpy.org/doc/stable/reference/generated/numpy.sum.html), and [matplotlib.pyplot.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

```python
import numpy as np
import matplotlib.pyplot as plt

n_trials = 1000
flips = np.random.randint(0, 2, size=n_trials)
cumulative_heads = np.cumsum(flips)
n = np.arange(1, n_trials + 1)
relative_freq = cumulative_heads / n

plt.plot(n, relative_freq)
plt.axhline(y=0.5, color="red", linestyle="--", label="Theoretical probability (0.5)")
plt.xlabel("Number of trials")
plt.ylabel("Relative frequency of heads")
plt.title("Coin Flip Simulation: Relative Frequency vs Number of Trials")
plt.legend()
plt.show()

print(f"Final estimated probability after {n_trials} trials: {relative_freq[-1]:.3f}")
```

2. **Experiment:** Rolling a fair die, **Event:** A number greater than 3 occurs.

```python
n_trials = 1000
rolls = np.random.randint(1, 7, size=n_trials)
event = (rolls > 3).astype(int)
cumulative = np.cumsum(event)
n = np.arange(1, n_trials + 1)
relative_freq = cumulative / n

plt.plot(n, relative_freq)
plt.axhline(y=0.5, color="red", linestyle="--", label="Theoretical probability (0.5)")
plt.xlabel("Number of trials")
plt.ylabel("Relative frequency")
plt.title("Rolling a Die: P(number > 3)")
plt.legend()
plt.show()
```

3. **Experiment:** Rolling two fair dice, **Event:** Sum of two dice equal 5.

```python
n_trials = 1000
die1 = np.random.randint(1, 7, size=n_trials)
die2 = np.random.randint(1, 7, size=n_trials)
event = (die1 + die2 == 5).astype(int)
cumulative = np.cumsum(event)
n = np.arange(1, n_trials + 1)
relative_freq = cumulative / n

plt.plot(n, relative_freq)
plt.axhline(y=4/36, color="red", linestyle="--", label="Theoretical probability (4/36)")
plt.xlabel("Number of trials")
plt.ylabel("Relative frequency")
plt.title("Rolling Two Dice: P(sum = 5)")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Number of rolls
n = 1000

# Simulate rolling two dice n times
die1 = np.random.randint(1, 7, size=n)
die2 = np.random.randint(1, 7, size=n)

# Calculate the sum of each pair
sums = die1 + die2

# Mark 1 where sum == 5, else 0
sum_is_5 = np.where(sums == 5, 1, 0)

# Cumulative count of successful outcomes
cumulative_successes = np.cumsum(sum_is_5)

# Trial numbers
trials = np.arange(1, n + 1)

# Estimated probability at each trial
estimated_prob = cumulative_successes / trials

# Plotting the convergence
plt.figure(figsize=(10, 5))
plt.plot(trials, estimated_prob)
plt.axhline(y=4/36, color='red', linestyle='--', label="True P(Sum = 5) = 4/36 ≈ 0.111")
plt.xlabel("Number of Trials")
plt.ylabel("Estimated Probability (Sum = 5)")
plt.title("Convergence of Estimated Probability: Sum of Two Dice = 5")
plt.legend()
plt.grid(True)
plt.show()
```

# Problem 2

Simulate the experiment of tossing $1000$ times a coin with probability of **head (coded as $1$) equal to $0.4$** and probability of **tail (coded as $0$) equal to $0.6$**. How can you check if the simulation of the experiment of been running as expected.

**Hint**: Use [matplotlib.pyplot.hist¶
](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html)

```python
n_trials = 1000
flips = np.random.choice([0, 1], size=n_trials, p=[0.6, 0.4])

plt.hist(flips, bins=[-0.5, 0.5, 1.5], rwidth=0.5)
plt.xticks([0, 1], ["Tail (0)", "Head (1)"])
plt.ylabel("Frequency")
plt.title("Simulated Coin Tosses (P(head)=0.4)")
plt.show()

p_head_simulated = np.mean(flips)
print(f"Simulated P(head) = {p_head_simulated:.3f}")
print(f"Expected P(head) = 0.4")
```
